```python
import numpy as np

def equation_49_hit_sum(
    x: np.ndarray,
    y: np.ndarray
) -> np.ndarray:
    """
    Implements Equation 49 (An) as the sum, for each pixel, of the minimum
    value (hit) between x and y at each time step.

    Args:
        x: np.ndarray, shape (T+1, N)
            First time series, values for each pixel and time
        y: np.ndarray, shape (T+1, N)
            Second time series, values for each pixel and time
    Returns:
        np.ndarray, shape (N,)
            Sum of hits for each pixel
    """
    # Calculate hits as minimum at each time and pixel
    hits = np.minimum(x, y)
    # Sum hits over all time steps for each pixel
    an = np.sum(
        hits,
        axis=0
    )
    return an

# Example usage with your previous arrays
x = np.array([
    [2, 5],
    [0, 4],
    [5, 1],
])
y = np.array([
    [4, 1],
    [1, 5],
    [0, 3],
])

an_result = equation_49_hit_sum(
    x,
    y
)
print(
    "Equation 49 results (An) for each pixel (sum of hits):",
    an_result
)
```

```python
import numpy as np

def equation_50_difference_sum(
    x: np.ndarray,
    y: np.ndarray
) -> np.ndarray:
    """
    Implements Equation 50 (Bn): for each pixel, sums (y - x) over all time points.
    Args:
        x: np.ndarray, shape (T+1, N)
            Time series for reference/classification 1
        y: np.ndarray, shape (T+1, N)
            Time series for reference/classification 2
    Returns:
        np.ndarray, shape (N,)
            Difference sum for each pixel
    """
    diff = y - x
    bn = np.sum(
        diff,
        axis=0
    )
    return bn

# Example usage with your previous arrays
x = np.array([
    [2, 5],
    [0, 4],
    [5, 1],
])
y = np.array([
    [4, 1],
    [1, 5],
    [0, 3],
])

bn_result = equation_50_difference_sum(
    x,
    y
)
print(
    "Equation 50 results (Bn) for each pixel:",
    bn_result
)
```

```python
import numpy as np

def equation_51_presence_temporal_diff(
    x: np.ndarray,
    y: np.ndarray
) -> np.ndarray:
    """
    Implements Equation 51 (Cn) from the referenced manuscript to 
    compare two time series of maps.
    For each pixel n, returns Cn as the temporal allocation difference 
    for presence, i.e., the sum of absolute differences between x and y 
    for all time points, minus the absolute difference of the summed 
    presences across time.
    
    Args:
        x: np.ndarray, shape (T+1, N)
            time series for reference/classification 1
        y: np.ndarray, shape (T+1, N)
            time series for reference/classification 2

    Returns:
        np.ndarray, shape (N,)
            value of Cn for each pixel
    """
    # Verify that both input arrays have same shape
    assert x.shape == y.shape, "x and y must have the same shape"
    
    # Calculate the presence difference at each time point (Pytn - Pxtn)
    diff = y - x
    
    # Sum the absolute differences over time points, for each pixel n
    abs_diff_sum = np.sum(
        np.abs(diff),
        axis=0
    )
    
    # Calculate the total presence for y and x over all time points, for each pixel
    total_y = np.sum(
        y,
        axis=0
    )
    total_x = np.sum(
        x,
        axis=0
    )
    
    # Bn is the absolute difference in total presence over time for each pixel
    bn = np.abs(
        total_y - total_x
    )
    
    # Equation 51: Cn = abs_diff_sum - Bn
    cn = abs_diff_sum - bn
    return cn

# Example data from your message
x = np.array(
    [
        [2, 5],
        [0, 4],
        [5, 1],
    ]
)

y = np.array(
    [
        [4, 1],
        [1, 5],
        [0, 3],
    ]
)

# Apply the function to x and y
equation_51_results = equation_51_presence_temporal_diff(
    x,
    y
)
print(
    "Equation 51 results (Cn) for each pixel:",
    equation_51_results
)
```

```python
import numpy as np

def equation_52_change_hits(
    x: np.ndarray,
    y: np.ndarray
) -> np.ndarray:
    """
    Implements Equation 52 (Dn) for any data type: counts the number of intervals
    where both x and y agree on the type of change (gain or loss) for each pixel.
    Args:
        x: np.ndarray, shape (T+1, N)
            First time series (any values)
        y: np.ndarray, shape (T+1, N)
            Second time series (any values)
    Returns:
        np.ndarray, shape (N,)
            Number of change hits for each pixel
    """
    # Calculate change (difference between consecutive time points)
    dx = x[1:, :] - x[:-1, :]
    dy = y[1:, :] - y[:-1, :]

    # Gain hit: both change positively (gain)
    gain_hit = ((dx > 0) & (dy > 0)).astype(int)
    # Loss hit: both change negatively (loss)
    loss_hit = ((dx < 0) & (dy < 0)).astype(int)

    # Dn: sum of gain and loss hits for each pixel
    dn = np.sum(gain_hit + loss_hit, axis=0)
    return dn

# Example data from your message
x = np.array([
    [2, 5],
    [0, 4],
    [5, 1],
])

y = np.array([
    [4, 1],
    [1, 5],
    [0, 3],
])

dn_result = equation_52_change_hits(
    x,
    y
)
print(
    "Equation 52 results (Dn) for each pixel (change hits):",
    dn_result
)
```

```python
import numpy as np

def equation_53_extent_difference(
    x: np.ndarray,
    y: np.ndarray
) -> np.ndarray:
    """
    Implements Equation 53 (En) from the referenced manuscript.
    For each pixel n, returns En as the difference in change (extent) 
    between y and x from the first to the last time point.
    
    Args:
        x: np.ndarray, shape (T+1, N)
            time series for reference/classification 1
        y: np.ndarray, shape (T+1, N)
            time series for reference/classification 2

    Returns:
        np.ndarray, shape (N,)
            value of En for each pixel
    """
    # Calculate the change in y from first to last time point for each pixel
    delta_y = y[-1, :] - y[0, :]
    # Calculate the change in x from first to last time point for each pixel
    delta_x = x[-1, :] - x[0, :]
    # Equation 53: En = (y_T - y_0) - (x_T - x_0)
    en = delta_y - delta_x
    return en

# Example data from your message
x = np.array(
    [
        [2, 5],
        [0, 4],
        [5, 1],
    ]
)

y = np.array(
    [
        [4, 1],
        [1, 5],
        [0, 3],
    ]
)

# Apply the function to x and y
equation_53_results = equation_53_extent_difference(
    x,
    y
)
print(
    "Equation 53 results (En) for each pixel:",
    equation_53_results
)
```

```python
import numpy as np

def equation_54_flexibility_difference(
    x: np.ndarray,
    y: np.ndarray
) -> np.ndarray:
    """
    Implements Equation 54 (Fn) from the referenced manuscript.
    For each pixel n, sums the absolute difference over all time steps between
    the changes in y and x from t-1 to t, then subtracts the absolute value of En.
    Args:
        x: np.ndarray, shape (T+1, N)
            time series for reference/classification 1
        y: np.ndarray, shape (T+1, N)
            time series for reference/classification 2
    Returns:
        np.ndarray, shape (N,)
            value of Fn for each pixel
    """
    # Calculate transitions for y and x
    dy = y[1:, :] - y[:-1, :]
    dx = x[1:, :] - x[:-1, :]
    # Difference of transitions, per time step and pixel
    abs_diff_transition = np.abs(dy - dx)
    # Sum over all time steps for each pixel
    sum_abs_diff = np.sum(abs_diff_transition, axis=0)

    # Calculate En using the function you already have
    en = equation_53_extent_difference(x, y)

    # Apply equation 54: Fn = summed transitions difference - |En|
    fn = sum_abs_diff - np.abs(en)
    return fn

# Example usage with your data
x = np.array(
    [
        [2, 5],
        [0, 4],
        [5, 1],
    ]
)
y = np.array(
    [
        [4, 1],
        [1, 5],
        [0, 3],
    ]
)

# Ensure you have the equation_53_extent_difference function defined as above!
equation_54_results = equation_54_flexibility_difference(
    x,
    y
)
print(
    "Equation 54 results (Fn) for each pixel:",
    equation_54_results
)
```
